```python
from tensorflow.keras.datasets import boston_housing
# (train_data, train_targets), (test_data, test_targets) = boston_housing
```

```python
(train_data, train_targets), (test_data, test_targets) = boston_housing.load_data()
```

```python
train_data.shape, test_data.shape, train_targets.shape
```

### Standarizing the dataset

```python
tr_mean = train_data.mean(axis = 0)
train_data -= tr_mean
tr_std = train_data.std(axis=0)
train_data /= tr_std
te_mean = test_data.mean(axis =0)
test_data -=te_mean
te_std = test_data.std(axis=0)
test_data /= te_std
```

```python
train_data.shape, test_data.shape
```

```python
from keras import models
from keras import layers
```

#### Model definition

```python
def build_model():
    model = models.Sequential()
    model.add(layers.Dense(64, activation = 'relu', input_shape = (train_data.shape[1],)))
    model.add(layers.Dense(64, activation= 'relu'))
    model.add(layers.Dense(1))
    model.compile(optimizer = 'rmsprop', loss = 'mse', metrics = ['mae'])
    return model
```

##### K-fold validation

```python
import numpy as np
k = 4 
num_val_samples = len(train_data) //k
num_epochs = 100 
all_scores = []
```

```python
for i in range(k):
    print('processing fold #', i)
    val_data = train_data[i * num_val_samples:(i+1) * num_val_samples]
    val_targets = train_targets[i *num_val_samples: (i+1) * num_val_samples]
    partial_train_data = np.concatenate([train_data[:i * num_val_samples], train_data[(i+1) * num_val_samples:]], axis = 0)
    partial_train_targets = np.concatenate([train_targets[:i*num_val_samples], train_targets[(i+1)* num_val_samples:]], axis =0 )
    
```

```python
model = build_model()
model.fit(partial_train_data, partial_train_targets, epochs = num_epochs, batch_size = 1, verbose = 0)
```
